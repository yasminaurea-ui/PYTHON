```python
import requests
import pandas as pd

# Indicador de crescimento do PIB
indicador = "NY.GDP.MKTP.KD.ZG"

# Países e regiões
paises = "WLD;USA;EUU;CHN"

# Link do Banco Mundial
url = f"https://api.worldbank.org/v2/country/{paises}/indicator/{indicador}"

# Período da pesquisa
parametros = {
    "format": "json",
    "date": "2015:2025",
    "per_page": 1000
}

# Buscar os dados
resposta = requests.get(url, params=parametros)
dados = resposta.json()[1]

# Organizar os dados
lista = []

for item in dados:
    if item["value"] is not None:
        lista.append({
            "País": item["country"]["value"],
            "Ano": item["date"],
            "Crescimento (%)": item["value"]
        })

# Criar tabela
tabela = pd.DataFrame(lista)

# Organizar os anos em colunas
tabela = tabela.pivot(
    index="País",
    columns="Ano",
    values="Crescimento (%)"
)

# Arredondar os valores
tabela = tabela.round(2)

# Mostrar tabela
print("CRESCIMENTO ECONÔMICO MUNDIAL (%)")
display(tabela)

# Salvar arquivo
tabela.to_csv(
    "tabela_crescimento_economico.csv",
    encoding="utf-8-sig"
)

print("Tabela criada com sucesso!")
```
